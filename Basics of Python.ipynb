```python
# complex numbers---> the part with j is imaginary number
z1 = 3 + 4j # we use 'j' instead of 'i', came from electrical engineering where 'i' is used for current
z2 = 2.5j  # Pure imaginary number (real part defaults to 0.0)

# Using the constructor
z3 = complex(3, 4)  # Equivalent to 3 + 4j

print(z1)
print(z2)
print(z3)

z = 5 - 2j

print(z.real)       # Output: 5.0
print(z.imag)       # Output: -2.0
print(z.conjugate()) # Output: (5+2j) 
# complex conjugate--> keeps the real part exactly the same and flips the sign of the imaginary part
```

```python
# operations on complex numbers
a = 2 + 3j
b = 1 - 1j

# Addition & Subtraction
print(a + b)  # Output: (3+2j)

# Multiplication & Division
print(a * b)  # Output: (5+1j)
print(a / b)  # Output: (-0.5+2.5j)

# Exponents
print(a ** 2) # Output: (-5+12j)
```

```python
'''
Note: Complex numbers cannot be compared using ordering operators like <, >, <=, or >=.
Doing so throws a TypeError. You can only check for equality (== or !=).
'''
print(a!=b)
print(a==b)
```

```python
first_name="Adnan"
second_name="Ahmad"
c='d'
print('The first name is {} and the last name is {} and {}'.format(first_name,second_name,c))
print('The first name is {} and the last name is {}'.format(second_name,first_name)) # order changed

# order can be restored using indexing
print('The first name is {1} and the last name is {0}'.format(second_name,first_name)) 

print('The first name is {b} and the last name is {a}'.format(a=second_name,b=first_name,))
```
